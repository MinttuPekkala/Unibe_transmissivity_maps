```python
a1 = np.random.choice([True, False], size=(4, 4), p=[0.1, 0.9])
```

```python
a2 = np.random.choice([True, False], size=(8, 8), p=[0.7, 0.3])
```

```python
plt.imshow(a1, cmap='gray')
```

```python
plt.imshow(a2, cmap='gray')
```

```python
h = min(a1.shape[0], a2.shape[0])
w = min(a1.shape[1], a2.shape[1])

arr1 = a1[:h, :w]
arr2 = a2[:h, :w]

diff = arr1 != arr2
print(diff)
```

```python
scale = 2

a1_small = a1.reshape(
    a2.shape[0], scale,
    a2.shape[1], scale
).mean(axis=(1, 3))
```

```python
plt.imshow(diff, cmap='gray')
plt.title("Differences")
#plt.colorbar(label="Value")

cbar = plt.colorbar(ticks=[0.0, 1.0])
cbar.ax.set_yticklabels(["Same", "Different"])
plt.show()
```

```python
mask = np.ma.masked_where(~diff, diff)

plt.imshow(arr1, cmap="gray")
plt.imshow(mask, cmap="autumn", alpha=0.8)

plt.title("Differences Highlighted")
plt.show()
```

```python
num_different = np.sum(diff)
```
